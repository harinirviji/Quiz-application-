```python

import csv
import logging
from datetime import datetime

# Logging configuration
logging.basicConfig(
    filename="quiz.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

# Quiz questions
questions = [
    {
        "question": "1. What is Python?",
        "options": ["A. Programming Language", "B. Browser", "C. Operating System", "D. Hardware"],
        "answer": "A"
    },
    {
        "question": "2. Which function displays output?",
        "options": ["A. input()", "B. print()", "C. type()", "D. int()"],
        "answer": "B"
    },
    {
        "question": "3. Which symbol is used for comments?",
        "options": ["A. //", "B. /*", "C. #", "D. &"],
        "answer": "C"
    },
    {
        "question": "4. Which data type stores text?",
        "options": ["A. int", "B. str", "C. float", "D. bool"],
        "answer": "B"
    },
    {
        "question": "5. Which keyword defines a function?",
        "options": ["A. function", "B. define", "C. def", "D. fun"],
        "answer": "C"
    },
    {
        "question": "6. Which loop repeats while a condition is true?",
        "options": ["A. if", "B. while", "C. break", "D. return"],
        "answer": "B"
    },
    {
        "question": "7. Which operator checks equality?",
        "options": ["A. =", "B. !=", "C. ==", "D. >"],
        "answer": "C"
    },
    {
        "question": "8. Which data type stores True or False?",
        "options": ["A. str", "B. bool", "C. list", "D. int"],
        "answer": "B"
    },
    {
        "question": "9. Which method adds an item to a list?",
        "options": ["A. add()", "B. insert()", "C. append()", "D. remove()"],
        "answer": "C"
    },
    {
        "question": "10. Which keyword handles exceptions?",
        "options": ["A. try", "B. print", "C. for", "D. class"],
        "answer": "A"
    }
]


def save_result(name, score):
    try:
        with open("quiz_results.csv", "a", newline="") as file:
            writer = csv.writer(file)
            writer.writerow([name, score, len(questions),
                             datetime.now().strftime("%Y-%m-%d %H:%M:%S")])
        logging.info("Quiz result saved successfully")

    except OSError as error:
        print("Unable to save result:", error)
        logging.error("Saving failed: %s", error)


def start_quiz():
    print("=" * 40)
    print("       PYTHON QUIZ APPLICATION")
    print("=" * 40)

    name = input("Enter your name: ").strip()

    if not name:
        name = "Student"

    score = 0

    for q in questions:
        print("\n" + q["question"])

        for option in q["options"]:
            print(option)

        while True:
            answer = input("Enter your answer (A/B/C/D): ").strip().upper()

            if answer in ["A", "B", "C", "D"]:
                break

            print("Invalid input! Please enter A, B, C or D.")

        if answer == q["answer"]:
            print("Correct answer!")
            score += 1
        else:
            print("Wrong answer! Correct answer:", q["answer"])

    print("\n" + "=" * 40)
    print("             QUIZ RESULT")
    print("=" * 40)

    print("Student Name:", name)
    print("Total Questions:", len(questions))
    print("Correct Answers:", score)
    print("Wrong Answers:", len(questions) - score)
    print("Final Score:", str(score) + "/" + str(len(questions)))

    percentage = (score / len(questions)) * 100
    print("Percentage:", round(percentage, 2), "%")

    save_result(name, score)

    print("\nThank you for participating!")


start_quiz()
```

```python

%%writefile app.py
import streamlit as st

st.set_page_config(
    page_title="Python Quiz Application",
    page_icon="🐍",
    layout="centered"
)

st.title("🐍 Python Quiz Application")
st.write("Test your Python programming knowledge!")

questions = [
    {
        "question": "1. Which language is Python?",
        "options": ["A. Programming language", "B. Markup language", "C. Database", "D. Operating system"],
        "answer": "A. Programming language"
    },
    {
        "question": "2. Which function is used to display output?",
        "options": ["A. input()", "B. print()", "C. display()", "D. show()"],
        "answer": "B. print()"
    },
    {
        "question": "3. Which symbol is used for comments in Python?",
        "options": ["A. //", "B. <!-- -->", "C. #", "D. **"],
        "answer": "C. #"
    },
    {
        "question": "4. Which data type stores whole numbers?",
        "options": ["A. float", "B. int", "C. str", "D. list"],
        "answer": "B. int"
    },
    {
        "question": "5. Which keyword defines a function?",
        "options": ["A. function", "B. define", "C. def", "D. fun"],
        "answer": "C. def"
    },
    {
        "question": "6. Which data type stores True or False?",
        "options": ["A. int", "B. bool", "C. str", "D. list"],
        "answer": "B. bool"
    },
    {
        "question": "7. Which operator is used for multiplication?",
        "options": ["A. +", "B. -", "C. *", "D. /"],
        "answer": "C. *"
    },
    {
        "question": "8. Which function gets input from the user?",
        "options": ["A. input()", "B. print()", "C. type()", "D. len()"],
        "answer": "A. input()"
    },
    {
        "question": "9. Which brackets are used to create a list?",
        "options": ["A. ()", "B. {}", "C. []", "D. <>"],
        "answer": "C. []"
    },
    {
        "question": "10. Which keyword is used for a decision?",
        "options": ["A. if", "B. repeat", "C. print", "D. input"],
        "answer": "A. if"
    }
]

if "submitted" not in st.session_state:
    st.session_state.submitted = False

with st.form("quiz_form"):
    answers = []

    for i, q in enumerate(questions):
        st.subheader(q["question"])
        choice = st.radio(
            "Choose your answer:",
            q["options"],
            key=f"question_{i}"
        )
        answers.append(choice)

    submitted = st.form_submit_button("Submit Quiz")

if submitted:
    score = 0

    for i, q in enumerate(questions):
        if answers[i] == q["answer"]:
            score += 1

    st.success("Quiz submitted successfully!")
    st.metric("Your Score", f"{score} / {len(questions)}")

    if score >= 8:
        st.balloons()
        st.write("Excellent work!")
    elif score >= 5:
        st.write("Good effort! Keep practising.")
    else:
        st.write("Keep learning and try again!")

    st.write("### Answer Review")

    for i, q in enumerate(questions):
        st.write(f"**{q['question']}**")
        st.write(f"Correct answer: {q['answer']}")
        st.write(f"Your answer: {answers[i]}")
```
